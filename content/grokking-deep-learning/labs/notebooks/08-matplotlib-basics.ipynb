# Matplotlib 小专题

目标：能够画出数据散点、训练曲线，并读懂图像中的趋势。正则化实验中的 `train_loss` 和 `validation_loss` 就会用到这些方法。

## 1. 基本对象

最常用的两个对象是：

- `fig`：整张画布；
- `ax`：画布中的一个坐标区域。

推荐使用面向对象写法：`fig, ax = plt.subplots()`。

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(x, y, label='sin(x)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('A line plot')
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```

## 2. 散点图：观察样本

`plot` 适合连续曲线；`scatter` 适合观察离散样本。正则化实验中的训练样本和验证样本通常应该先画出来，确认数据切分是否合理。

```python
rng = np.random.default_rng(0)
x = np.linspace(-1, 1, 30)
y = np.sin(3 * x) + rng.normal(0, 0.15, size=len(x))

fig, ax = plt.subplots(figsize=(7, 4))
ax.scatter(x, y, label='samples')
ax.plot(x, np.sin(3 * x), color='black', label='clean signal')
ax.set_xlabel('x')
ax.set_ylabel('target')
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```

## 3. 训练曲线

训练曲线的横轴通常是 epoch，纵轴是 loss。多条曲线放在同一个 `ax` 上，用 `label` 和 `legend` 区分。

```python
epochs = np.arange(100)
train_loss = np.exp(-epochs / 25)
validation_loss = 0.2 + 0.8 * np.exp(-epochs / 15)

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(epochs, train_loss, label='train loss')
ax.plot(epochs, validation_loss, label='validation loss')
ax.set_xlabel('epoch')
ax.set_ylabel('loss')
ax.set_title('Training curves')
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```
